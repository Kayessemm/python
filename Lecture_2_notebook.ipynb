```python
# Else statement tells python what code to run if the if statement proves false

year = 2001

if year >= 2000:
    print("We are in the 21st century")
else:
    print('We are not in the 21st century')
```

```python
year = 1867

if year >= 2000:
    print('We are in the 21st century')
elif year >= 1900:
    print('We are in the 20th century')
elif year >= 1800:
    print('We are in the 19th century.')
else:
    print('We are way back in time!')
```
